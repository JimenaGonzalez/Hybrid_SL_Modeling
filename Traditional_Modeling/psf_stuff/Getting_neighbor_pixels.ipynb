```python
import healpy as hp
import numpy as np
import pandas as pd
```

```python
def hpix_disc(ra, dec, nside=16384, radius_arcmin=3.0, inclusive=True):
    """
    Return the HEALPix (NEST) pixel indices within radius_arcmin
    of (ra, dec), at the given nside.
    """
    theta = np.radians(90.0 - dec)
    phi = np.radians(ra)
    vec = hp.ang2vec(theta, phi)
    radius_rad = np.radians(radius_arcmin / 60.0)
    return hp.query_disc(nside, vec, radius_rad, nest=True, inclusive=inclusive)
```

```python
pathdata = '/Users/jimenagonzalez/research/DSPL/Hybrid_SL_Modeling/data/candidates_data/'
filename_data = 'sledcand_old.csv'
data = pd.read_csv(pathdata + filename_data)

data.head()
```

```python
all_pix = set()
for ra, dec in zip(data['RA'], data['DEC']):
    all_pix.update(hpix_disc(ra, dec, nside=16384, radius_arcmin=5.0))

hpix_table = pd.DataFrame({'hpix_4096': sorted(all_pix)})
hpix_table.to_csv('stars_data/hpixstars_sledcand.csv', index=False)
```
